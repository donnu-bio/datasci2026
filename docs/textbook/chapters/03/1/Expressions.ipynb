# Вирази (Expressions)

Мови програмування набагато простіші за людські мови. Проте в будь-якій мові є певні граматичні правила, які потрібно вивчити, і саме з цього ми й почнемо. У цьому тексті ми використовуватимемо мову програмування [Python](https://www.python.org/). Вивчення граматичних правил є надзвичайно важливим, адже ті самі правила, що застосовуються в найпростіших програмах, є також основою для більш складних програм.

Програми складаються з *виразів*, які описують комп’ютеру, як поєднувати фрагменти даних. Наприклад, вираз множення складається з символу `*` між двома числовими виразами. Вирази, такі як `3 * 4`, *обчислюються* комп’ютером. Значення (результат *обчислення*) останнього виразу в кожній комірці, у цьому випадку `12`, відображається під коміркою.

```python
3 * 4
```

Граматичні правила мови програмування є жорсткими. У Python символ `*` не може зустрічатися двічі поспіль. Комп’ютер не намагатиметься інтерпретувати вираз, який не відповідає попередньо заданим структурам виразів. Натомість він видасть помилку `SyntaxError`. *Синтаксис* (*Syntax*) мови — це набір її граматичних правил, а помилка `SyntaxError` вказує на те, що структура виразу не відповідає жодному з правил цієї мови.

```python
3 * * 4
```

Невеликі зміни у виразі можуть повністю змінити його значення. Нижче пробіл між символами `*` було видалено. Оскільки `**` стоїть між двома числовими виразами, цей вираз є правильно сформованим виразом *піднесення до степеня* (перше число, піднесене до степеня другого: 3 помножене на 3 помножене на 3 помножене на 3). Символи `*` та `**` називаються *операторами* (*operators*), а значення, які вони поєднують, — *операндами* (*operands*).

```python
3 ** 4
```

**Поширені оператори.** У науці про дані часто доводиться поєднувати числові значення, і набір операторів у мові програмування розроблено таким чином, щоб за допомогою виразів можна було виражати будь-які арифметичні операції. У Python найважливішими є такі оператори.

| Тип виразу | Оператор | Приклад    | Значення     |
|-----------------|----------|------------|-----------|
| Додавання        | `+`      | `2 + 3`    | `5`       |
| Віднімання     | `-`      | `2 - 3`    | `-1`      |
| Множення  | `*`      | `2 * 3`    | `6`       |
| Ділення        | `/`      | `7 / 3`    | `2,66667` |
| Залишок       | `%`      | `7 % 3`    | `1`       |
| Піднесення до степеня | `**`     | `2 ** 0,5` | `1,41421` |

Вирази в Python підпорядковуються тим самим звичним правилам *пріоритету*, що й в алгебрі: множення та ділення виконуються раніше, ніж додавання та віднімання. Дужки можна використовувати для об’єднання менших виразів у рамках більшого виразу.

```python
1 + 2 * 3 * 4 * 5 / 6 ** 3 + 7 + 8 - 9 + 10
```

```python
1 + 2 * (3 * 4 * 5 / 6) ** 3 + 7 + 8 - 9 + 10
```

У цьому розділі представлено багато типів виразів. Навчання програмуванню вимагає практичного застосування всього вивченого в різних комбінаціях та дослідження поведінки комп’ютера. Що станеться, якщо поділити на нуль? Що станеться, якщо поділити двічі поспіль? Не завжди потрібно звертатися до експерта (або до Інтернету); багато з цих деталей можна з’ясувати, спробувавши самостійно.
