# Приклад: Темпи зростання (Growth Rates)

Взаємозв’язок між двома значеннями однієї й тієї ж величини, виміряними в різний час, часто позначають як *темп зростання* (*growth rate*). Наприклад, у 2002 році федеральний уряд США працевлаштував 2 766 000 осіб, а у 2012 році — 2 814 000 осіб. Щоб обчислити темп зростання, спочатку потрібно визначитися, яке значення вважати «початковим». Для значень, що змінюються з часом, природним вибором є попереднє значення. Потім різницю між «зміненим» і «початковим» значеннями ділимо на «початкове» значення.

```python
initial = 2766000
changed = 2814000
(changed - initial) / initial
```

Також можна відняти одиницю від відношення цих двох вимірювань, що дасть такий самий результат.

```python
(changed/initial) - 1
```

Це значення відображає темп зростання за 10 років. Корисною властивістю темпів зростання є те, що вони не змінюються, навіть якщо значення вказані в інших одиницях виміру. Так, наприклад, ми можемо відобразити ту саму залежність між кількістю тисяч людей у 2002 та 2012 роках.

```python
initial = 2766
changed = 2814
(changed/initial) - 1
```

За 10 років кількість працівників федерального уряду США зросла лише на 1,74 %. За цей час загальні видатки федерального уряду США зросли з 2,37 трлн доларів до 3,38 трлн доларів у 2012 році.

```python
initial = 2.37
changed = 3.38
(changed/initial) - 1
```

Зростання федерального бюджету на 42,6 % значно перевищує приріст чисельності федеральних службовців, який склав 1,74 %. Насправді чисельність федеральних службовців зростала набагато повільніше, ніж населення Сполучених Штатів, яке за той самий період збільшилося на 9,21 % — з 287,6 млн осіб у 2002 році до 314,1 млн у 2012 році.

```python
initial = 287.6
changed = 314.1
(changed/initial) - 1
```

Темп зростання може бути від’ємним, що свідчить про зменшення певного показника. Наприклад, кількість робочих місць у виробничій галузі США зменшилася з 15,3 млн у 2002 році до 11,9 млн у 2012 році, що відповідає темпу зростання -22,2 %.

```python
initial = 15.3
changed = 11.9
(changed/initial) - 1
```

Річний темп зростання — це темп зростання певної величини протягом одного року. Річний темп зростання, що становить 0,035, що накопичується щороку протягом 10 років, дає набагато більший десятирічний темп зростання — 0,41 (або 41 %).

```python
1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 - 1
```

Те саме обчислення можна записати, використовуючи імена та степені.

```python
annual_growth_rate = 0.035
ten_year_growth_rate = (1 + annual_growth_rate) ** 10 - 1
ten_year_growth_rate
```

Аналогічно, десятирічний темп зростання можна використовувати для обчислення еквівалентного річного темпу зростання. Нижче `t` — це кількість років, що минули між вимірюваннями. У наведеному нижче прикладі обчислюється річний темп зростання федеральних видатків за останні 10 років.

```python
initial = 2.37
changed = 3.38
t = 10
(changed/initial) ** (1/t) - 1
```

Загальне зростання за 10 років відповідає щорічному приросту на 3,6 %.

Отже, коефіцієнт зростання `g` використовується для опису відносного розміру `initial` величини та `changed` величини через певний проміжок часу `t`. Щоб обчислити `changed` величину, потрібно `t` разів послідовно застосовувати коефіцієнт зростання `g` за допомогою піднесення до степеня.

`initial * (1 + g) ** t`

Щоб обчислити `g`, піднесіть загальний приріст до степеня `1/t` і відніміть одиницю.

`(changed/initial) ** (1/t) - 1`
