# Імена

У Python значенням присвоюються імена за допомогою оператора *присвоєння*. У операторі присвоєння за іменем йде знак `=`, за яким слідує будь-який вираз. Значення виразу, що стоїть праворуч від `=`, *присвоюється* цьому імені. Після того як імені присвоєно значення, у подальших виразах це значення замінюватиме це ім’я.

```python
a = 10
b = 20
a + b
```

У виразі праворуч від `=` можна використовувати раніше присвоєне ім’я.

```python
quarter = 1/4
half = 2 * quarter
half
```

Однак імені присвоюється лише поточне значення виразу. Якщо це значення згодом зміниться, імена, які були визначені на основі цього значення, не зміняться автоматично.

```python
quarter = 4
half
```

Імена повинні починатися з літери, але можуть містити як літери, так і цифри. Ім’я не може містити пробілів; замість цього зазвичай використовують символ підкреслення `_` для заміни кожного пробілу. Користь імен залежить від того, як ви їх використовуєте; саме програміст повинен обирати імена, які легко інтерпретувати. Зазвичай можна придумати більш змістовні імена, ніж `a` та `b`. Наприклад, щоб описати податок з продажу на покупку вартістю 5 доларів у Берклі, штат Каліфорнія, наведені нижче імена пояснюють значення різних величин, що беруть участь у розрахунку.

```python
purchase_price = 5
state_tax_rate = 0.075
county_tax_rate = 0.02
city_tax_rate = 0
sales_tax_rate = state_tax_rate + county_tax_rate + city_tax_rate
sales_tax = purchase_price * sales_tax_rate
sales_tax
```
