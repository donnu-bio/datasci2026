```python
from datascience import *
path_data = '../../../assets/data/'
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')

cones = Table.read_table(path_data + 'cones.csv')
nba = Table.read_table(path_data + 'nba_salaries.csv').relabeled(3, 'SALARY')
movies = Table.read_table(path_data + 'movies_by_year.csv')
```

Таблиці — це основний спосіб представлення наборів даних. Таблицю можна розглядати у двох напрямках:
* як послідовність іменованих стовпців, кожен з яких описує окрему характеристику всіх рядків у наборі даних, або
* як послідовність рядків, кожен з яких містить всю інформацію про окремого суб’єкта в наборі даних.

Ми детально вивчатимемо таблиці в наступних кількох розділах. А поки що ми лише ознайомимося з кількома методами, не заглиблюючись у технічні подробиці.

Таблиця `cones` вже імпортована для нас; пізніше ми побачимо, як саме, а тут просто попрацюємо з нею. Спочатку давайте поглянемо на неї.

```python
cones
```

Таблиця має шість рядків. Кожен рядок відповідає одному ріжку морозива. Ріжки морозива — це *індивідууми*.

Кожен ріжок має три атрибути: смак, колір і ціну. Кожен стовпець містить дані про один із цих атрибутів, тому всі записи в будь-якому стовпці є одного типу. Кожен стовпець має назву. Ми будемо посилатися на стовпці за їхніми назвами.

Метод (method) таблиці схожий на функцію, але він повинен оперувати таблицею. Тому виклик має вигляд

`назва_таблиці.метод(аргументи)`
`name_of_table.method(arguments)`

Наприклад, якщо ви хочете переглянути лише перші два рядки таблиці, ви можете використати метод таблиці `show`.

```python
cones.show(2)
```

Ви можете замінити 2 на будь-яку кількість рядків. Якщо ви запитаєте більше шести, ви отримаєте лише шість, оскільки `cones` має лише шість рядків.

## Вибір групи стовпців
Метод `select` створює нову таблицю, що складається виключно із зазначених стовпців.

```python
cones.select('Flavor')
```

При цьому вихідна таблиця залишається без змін.

```python
cones
```

Ви можете вибрати кілька стовпців, розділивши їхні назви комами.

```python
cones.select('Flavor', 'Price')
```

Ви також можете *видалити* (*drop*) непотрібні стовпці. Наведену вище таблицю можна створити, видаливши стовпець `Color`.

```python
cones.drop('Color')
```

Ви можете надати цій новій таблиці ім’я та переглянути її ще раз, просто ввівши її ім’я.

```python
no_colors = cones.drop('Color')

no_colors
```

Як і `select`, метод `drop` створює меншу таблицю, залишаючи вихідну таблицю без змін. Щоб дослідити ваші дані, ви можете створювати будь-яку кількість менших таблиць, додаючи або видаляючи стовпці. Це ніяк не зашкодить вашій вихідній таблиці даних.

## Сортування рядків

Метод `sort` створює нову таблицю, упорядковуючи рядки вихідної таблиці у порядку зростання (ascending) значень у вказаному стовпці. У цьому прикладі таблиця `cones` відсортована за ціною рожків у порядку зростання.

```python
cones.sort('Price')
```

Щоб відсортувати в порядку зменшення (descending), можна використати *опціональний* (*optional*) аргумент функції `sort`. Як випливає з назви, опціональні аргументи не є обов’язковими, але їх можна використовувати, якщо потрібно змінити поведінку методу за замовчуванням.

За замовчуванням функція `sort` сортує за зростанням значень у вказаному стовпці. Щоб відсортувати в порядку зменшення, використовуйте опціональний аргумент `descending=True`.

```python
cones.sort('Price', descending=True)
```

Як і `select` та `drop`, метод `sort` не змінює вихідну таблицю.

## Вибір рядків, що відповідають умові
Метод `where` створює нову таблицю, що складається виключно з рядків, які відповідають заданій умові. У цьому розділі ми будемо працювати з дуже простою умовою: значення у вказаному стовпці має дорівнювати значенню, яке ми також вказуємо. Отже, метод `where` має два аргументи.

```python
cones.where('Flavor', 'chocolate')
```

Аргументи, розділені комою, — це назва стовпця та значення, яке ми шукаємо в цьому стовпці. Метод `where` також можна використовувати, коли умова, якій повинні відповідати рядки, є складнішою. У таких випадках виклик методу також буде дещо складнішим.

Важливо вказати значення точно. Наприклад, якщо ми вкажемо `Chocolate` замість `chocolate`, то `where` правильно не знайде жодного рядка, де смак — `Chocolate`.

```python
cones.where('Flavor', 'Chocolate')
```

Як і всі інші табличні методи в цьому розділі, `where` залишає вихідну таблицю без змін.

## Приклад: зарплати в НБА

«NBA є найбільш високооплачуваною професійною спортивною лігою у світі», [повідомляло CNN](http://edition.cnn.com/2015/12/04/sport/gallery/highest-paid-nba-players/) у березні 2016 року. Таблиця «nba» містить [зарплати всіх Національних баскетбольних асоціацій гравців](https://www.statcrunch.com/app/index.php?dataid=1843341) у 2015-2016 рр.

Кожен рядок представляє одного гравця. Стовпці:
| **Мітка стовпця** | опис                                    |
|--------------------|-----------------------------------------------------|
| `PLAYER`           | Ім'я гравця                                       |
| `POSITION`         | Позиція гравця в команді                        |
| `TEAM`             | Назва команди                                       |
|`SALARY`    | Зарплата гравця в 2015-2016 роках, мільйони доларів|
 
Код позиції такий PG (Розігруючий захисник Point Guard), SG (Атакуючий захисник Shooting Guard), PF (Силовий форвард Power Forward), SF (малий форвард Small Forward), and C (Центровий Center). Однак у подальшому тексті не йдеться про те, як саме грають у баскетбол.

У першому рядку вказано, що Пол Міллсап, сильний форвард команди «Атланта Хокс», у сезоні 2015–2016 отримав зарплату в розмірі майже $\$18,7$ мільйонів.

```python
nba
```

Шанувальники Стівена Каррі можуть знайти його рядок за допомогою `where`.

```python
nba.where('PLAYER', 'Stephen Curry')
```

Ми також можемо створити нову таблицю під назвою `warriors`, яка міститиме лише дані про команду «Голден Стейт Ворріорс».

```python
warriors = nba.where('TEAM', 'Golden State Warriors')
warriors
```

За замовчуванням відображаються перші 10 рядків таблиці. Ви можете скористатися командою `show`, щоб відобразити більше або менше рядків. Щоб відобразити всю таблицю, скористайтеся командою `show` без аргументів у дужках.

```python
warriors.show()
```

Таблиця `nba` відсортована за алфавітом за назвами команд. Щоб побачити, скільки заробляли гравці у сезоні 2015–2016 років, доцільно відсортувати дані за розміром зарплати. Пам’ятайте, що за замовчуванням сортування здійснюється у порядку зростання.

```python
nba.sort('SALARY')
```

Ці цифри дещо складно порівнювати, оскільки деякі з цих гравців протягом сезону змінювали команди й отримували зарплату від більш ніж однієї команди; у таблиці вказано лише зарплату від останньої команди.

Репортаж CNN присвячений протилежному кінцю шкали зарплат — гравцям, які входили до числа найбільш високооплачуваних у світі. Щоб визначити цих гравців, можна відсортувати список за зарплатою у спадному порядку (descending order) й розглянути кілька верхніх рядків.

```python
nba.sort('SALARY', descending=True)
```
